# 1. Perceptron

```python
import pandas as pd

dataframe = pd.read_csv("C:/Users/HP/Desktop/Doctorat/Teaching/MLDL/KDD_Binary_classification.csv")
#print(dataframe.head())
print(dataframe.shape)

X = dataframe.drop(["label"], axis=1)
Y = dataframe["label"]

print("X shape:", X.shape)
print("Y shape:", Y.shape)

print("\nFeatures:")
print(X.head())                                                   #Display the first X lines in the dataframe. X.tail(x) for the last x lines.

print("\nTarget:")
print(Y.head())
```

```python
print(Y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)     #42 is the seed used for data shefling before divison

print(X_train.shape)
print(X_test.shape)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neural_network import MLPRegressor

model_Perceptron = make_pipeline(
    StandardScaler(),
    MLPRegressor(
        max_iter=1000,
        tol=1e-3,
        eta0=1.0,
        shuffle=True,
        random_state=42
    )
)

model_Perceptron.fit(X_train, Y_train)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,confusion_matrix, classification_report, roc_auc_score

Y_pred = model_Perceptron.predict(X_test)

print("Accuracy ↑:", accuracy_score(Y_test, Y_pred))
print("Precision ↑:", precision_score(Y_test, Y_pred, pos_label="attack"))
print("Recall ↑:", recall_score(Y_test, Y_pred, pos_label="attack"))
print("F1-score ↑:", f1_score(Y_test, Y_pred, pos_label="attack"))
print("Confusion matrix:")
print(confusion_matrix(Y_test, Y_pred))
```

```python
print(X_test[0:1],"\n")                    
Y_pred = model_Perceptron.predict(X_test[0:1])
print("Prediction:", Y_pred[0])

print("Target:", Y_test.iloc[0])  
```

# 2. Multi-Layer Perceptron (MLP)

## 2.1. Classification 

```python
import pandas as pd 
from sklearn.datasets import fetch_california_housing

housing=fetch_california_housing()
x=pd.DataFrame(housing.data,columns=housing.feature_names)
y=pd.Series(housing.target,name="MedHouseVal")

print("x shape :",x.shape)
print("y shape :",y.shape)

print("\n features:")
print(x.head())

print("\n target:")
print(y.head())
```

```python
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.2, random_state=42)     #42 is the seed used for data shefling before divison

print(X_train.shape)
print(X_test.shape)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

model_MLP = make_pipeline(
    StandardScaler(),
    MLPRegressor(
        hidden_layer_sizes=(50,25),
        activation="relu",
        solver="sgd",
        learning_rate_init=0.001,
        max_iter=1000,
        random_state=42
    )
)

model_MLP.fit(X_train, Y_train)
```

```python

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

Y_pred = model_MLP.predict(X_test)

print("Testing MAE ↓:", mean_absolute_error(Y_test, Y_pred))
print("Testing MSE ↓:", mean_squared_error(Y_test, Y_pred))
print("Testing RMSE ↓:", np.sqrt(mean_squared_error(Y_test, Y_pred)))
print("Testing R² ↑:", r2_score(Y_test, Y_pred))
print("")
```
